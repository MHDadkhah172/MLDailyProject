```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_df = pd.read_csv('../data/train.csv')
test_df = pd.read_csv('../data/test.csv')
```

```python
train_df.describe(include='all')
```

```python
test_df.describe(include='all')
```

```python
print(train_df.isnull().sum())
```

```python
features = [col for col in train_df.columns if col != 'price_range']

fig, axes = plt.subplots(nrows=5, ncols=4, figsize=(20, 22))
axes = axes.flatten()

for i, col in enumerate(features):
    sns.scatterplot(
        data=train_df,
        x=col,
        y='price_range',
        hue='price_range',
        palette='viridis',
        ax=axes[i],
        alpha=0.6,
        legend=False
    )
    axes[i].set_title(f'{col} vs Price Range', fontsize=11, fontweight='bold')
    axes[i].set_yticks([0, 1, 2, 3])
    axes[i].set_ylabel('Price Range')
    axes[i].set_xlabel(col)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16, 12))
corr_matrix = train_df.corr()

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    cbar_kws={'label': 'Correlation Coefficient'}
)

plt.title('Correlation Heatmap of Features and Target (price_range)', fontsize=14, pad=15)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

target_corr = train_df.corr()['price_range'].drop('price_range').sort_values(ascending=False)

sns.barplot(
    x=target_corr.values,
    y=target_corr.index,
    hue=target_corr.index,
    palette='magma',
    legend=False
)

plt.title('Correlation of Features with Target (price_range)', fontsize=13)
plt.xlabel('Correlation Coefficient')
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

features = [col for col in train_df.columns if col != 'price_range']

fig, axes = plt.subplots(nrows=5, ncols=4, figsize=(20, 22))
axes = axes.flatten()

for i, col in enumerate(features):
    if col == 'px_height':
        counts, bin_edges = np.histogram(train_df[col], bins=30)
        bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])
        width = bin_edges[1] - bin_edges[0]

        bars = axes[i].bar(bin_centers, counts, width=width * 0.9, edgecolor='black', alpha=0.85, color='#3498db')
        bars[0].set_facecolor('#e74c3c')

    elif train_df[col].nunique() <= 25:
        counts = train_df[col].value_counts().sort_index()
        x_ticks = [str(x) for x in counts.index]

        colors = ['#e74c3c' if idx == 0 and col == 'sc_w' else '#3498db' for idx in counts.index]

        axes[i].bar(x_ticks, counts.values, color=colors, edgecolor='black', alpha=0.85, width=0.6)

    else:
        sns.histplot(train_df[col], bins=30, kde=True, ax=axes[i], color='#3498db', edgecolor='black')

    axes[i].set_title(f'Distribution of {col}', fontsize=11, fontweight='bold')
    axes[i].set_xlabel(col, fontsize=10)
    axes[i].set_ylabel('Count', fontsize=10)
    axes[i].grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
train_df = train_df[(train_df['px_height'] > 0)].reset_index(drop=True)
train_df = train_df[(train_df['sc_w'] > 0)].reset_index(drop=True)
```

```python
train_df.to_csv('../data/train_cleaned.csv', index=False)
test_df.to_csv('../data/test_cleaned.csv', index=False)
```
